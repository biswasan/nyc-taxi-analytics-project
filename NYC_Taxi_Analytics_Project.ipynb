```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv"

df = pd.read_csv(url)

df.head()
```

```python
df_clean = df.dropna()

df_clean = df_clean[
    (df_clean["fare"] > 0) &
    (df_clean["distance"] > 0)
]

df_clean.head()
```

```python
summary = df_clean.groupby("pickup_borough").agg(
    total_rides=("fare", "count"),
    total_fare=("fare", "sum"),
    avg_fare=("fare", "mean"),
    avg_distance=("distance", "mean")
).reset_index()

summary
```

```python
df_clean.to_csv("clean_taxi_data.csv", index=False)
summary.to_csv("taxi_kpi_summary.csv", index=False)
```
